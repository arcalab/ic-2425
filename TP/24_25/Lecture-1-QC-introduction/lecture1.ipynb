```python
import pennylane as qml
import numpy as np
```

```python
n_qubits = 2 

dev = qml.device("default.qubit", wires=n_qubits)
```

```python
@qml.qnode(dev)
def circuito():

    for i in range(n_qubits):
        qml.Hadamard(wires=i)

    return qml.probs(wires=0)
```

```python
circuito()
```

```python
#qml.draw(circuito)()
qml.draw_mpl(circuito)()
```

```python
#para 5 qubits 

n_qubits = 1

dev = qml.device("default.qubit", wires=n_qubits)

@qml.qnode(dev)
def circuito():

    qml.RY(2*np.arcsin(np.sqrt(0.2)), wires=0)
    return qml.probs(wires=0)

print(circuito())
qml.draw_mpl(circuito)()
```
